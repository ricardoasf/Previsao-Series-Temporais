```python
#Definição das bibliotecas usadas
import pandas as pd #álgebra linear
import numpy as np #processamento de dados

import matplotlib as mpl
import matplotlib.pyplot as plt #visualização de dados
import seaborn as sns #visualização de dados

############### PARAMETRIZAÇÃO PARA EXECUÇÃO #############
# 1 - para Jupyter/Outras IDEs;
# 2 - para GoogleColab
carga_dados = 2
##########################################################

if carga_dados == 1:
    #Carrega Série Temporal a partir de Planilha de Dados no formato de DataFrame
    import tkinter as tk
    from tkinter import filedialog
    root = tk.Tk()
    root.withdraw()
    caminho_arquivo = filedialog.askopenfilename()
    df = pd.read_csv(caminho_arquivo)
else:
    #Carrega Dados para GoogleColab
    import io
    from google.colab import files
    uploaded = files.upload()
    #Obtém o nome do arquivo que foi feito upload
    nome_arquivo = next(iter(uploaded))
    df = pd.read_csv(io.BytesIO(uploaded[nome_arquivo]))

df
```

```python
#Importa as funções para calcular e plotar a ACF e a PACF
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf

#Cria uma figura de 12 x 4 polegadas com 1 linha e 2 colunas de gráficos
fig, ax = plt.subplots(1, 2, figsize=(12, 4))

#Plota a função de autocorrelação (ACF) com Lag = 20
plot_acf(df['Preco'], lags=20, ax=ax[0])
ax[0].set_title("ACF")

#Plota a função de autocorrelação parcial (PACF) com Lag = 20 e método Yule-Walker com ajuste de máxima verossimilhança
plot_pacf(df['Preco'], lags=20, ax=ax[1], method="ywm")
ax[1].set_title("PACF")

#Ajusta espaçamentos entre os gráficos
plt.tight_layout()

#Mostra Figura
plt.show()
```
